#Tipado en Python
Para eliminar la incosistencia en el codigo y poder hacerlo mas legible es necesario saber que son las anotaciones. Esto hace que podamos evitar bugs de interpretacion de variables.

##Anotation o type anotation
Son utilizados para definir cada tipo de valor para cada variable utilizadas en nuestro programa.

Estas nos ayudan en la lectura de codigo a los desarrolladores, pero al interprete no le otorgan nada.

Con el formato:

Nombre de variable : tipo = Valor

###Tipos basicos

```python
a: int = 12
b: str = "Anotacion de tipo"
c: bool = True
d: float = 4,12
```

###Tipos None

```python
e: None = None
```

###Listas, Tuplas y Diccionarios
Si nosotros solo queremos definir la variable ponemos:

Variable: tipo (list,tuple,dict,set) = Valores {} [] ()

Si nosotros tambien queremos colocar el tipo de valor que debe poseer cada estructura debemos importar de la libreria typing los modulos de los datos que vamos a tipear como List, Dict y Tuple, con la siguiente estructura:

variable: Tipo[tipo de los datos que integrara] = valores

```python
from typing import List, Tuple, Dict
lista: List[int] = [1,2,3]
diccionario: Dict[str , int] = {"Uno": 1, "Dos" : 2 }
tupla: Tuple[str,str] = ("One","Dos")
```

###Constantes
En python no existen las constantes pero con el modulo Final de la libreria typing podremos declarar a una variable como constante, esta variable siempre debe ir en mayuscula y seguida del modulo Final:

VARIABLE: Final[Tipo de variable] = Valores

```python
from typing import Final
PI: Final[float] = 3.1416
SETTINGS: Final = (True,"rt",23432)
```

###Union
Para declarar variables que van a cambiar de valor durante la ejecucion del programa usamos el modulo Union con los distintos tipos de valores que puede asignarse a la variable:

Variable: Union[valores que puede tomar la variable ] = Valor inicial

```python
from typing import Union
nombre: Union[None, str] = None
print (nombre)
nombre = "Carlos"
print (nombre)
```

```python
from typing import List
def average (numbers: List[int]) -> Union[None,float]:
  if len(numbers) == 0:
    return None
  else:
    return sum(numbers)/len(numbers)
numbers: List[int] = [4,5,6,2,7]
result: Union[None, float]= average(numbers)
print(result)
```

###Alias
Con los alias podemos reducir lineas de codigo en python para hacer mas legible nuestro codigo.
Los alias se escriben comensando con mayusculas y se utilizan como plantillas del tipado de las variables:
Variable: Tipo[Valores dentro la variable]

```python
from typing import Tuple, Union
Conection = Tuple[str,str,int]
def database_conection ( conection: Conection) -> Union[Tuple[str,str,int],None]:
  if conection[0] == "root":
    return conection
  else:
    return None
conection: Conection = ("root","2122",12345)
print(database_conection(conection))
```

###Tipo de dato self
Con este tipado declaramos que

```python
from typing import Self

class User:
  def __init__(self, name: str):
    self.name = name

  def set_name(self, new_name: str) -> Self:
    self.name = new_name
    return self

  def greet(self) -> str:
    return f"Hello, my name is {self.name}"


user_instance = User("Alice")
print(user_instance.greet())

# Method chaining using Self
user_instance_chained = User("Bob").set_name("Robert")
print(user_instance_chained.greet())
```

###Named tuple

```python
from typing import NamedTuple

class Point(NamedTuple):
    x: float
    y: float

# Create instances of the NamedTuple
point1: Point = Point(x=10.0, y=20.0)
point2: Point = Point(30.0, 40.0)

print(f"Point 1: x={point1.x}, y={point1.y}")
print(f"Point 2: x={point2[0]}, y={point2[1]}")

class Person(NamedTuple):
    name: str
    age: int
    city: str = "Unknown"

person1: Person = Person(name="Alice", age=30)
person2: Person = Person("Bob", 25, "New York")

print(f"Person 1: {person1.name}, {person1.age}, {person1.city}")
print(f"Person 2: {person2.name}, {person2.age}, {person2.city}")
```

###Type var

```python
from typing import TypeVar, List

T = TypeVar('T')

def first_element(items: List[T]) -> T:
    if not items:
        raise ValueError("List is empty")
    return items[0]

def identity(arg: T) -> T:
    return arg

int_list = [1, 2, 3]
first_int = first_element(int_list)
print(f"First integer: {first_int}, Type: {type(first_int)}")

str_list = ["apple", "banana", "cherry"]
first_str = first_element(str_list)
print(f"First string: {first_str}, Type: {type(first_str)}")

id_int = identity(123)
print(f"Identity of int: {id_int}, Type: {type(id_int)}")

id_str = identity("hello")
print(f"Identity of string: {id_str}, Type: {type(id_str)}")

Numeric = TypeVar('Numeric', int, float)

def add_two_numbers(a: Numeric, b: Numeric) -> Numeric:
    return a + b

print(f"Sum of two ints: {add_two_numbers(5, 10)}")
print(f"Sum of two floats: {add_two_numbers(5.5, 10.5)}")
```
